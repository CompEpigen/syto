```python
data_path = "../Data/Curated/MethylBERT_Figure3D_customsplit/"
experiment_name = "MethylBERT_Figure3D"
max_sequence_length = 150
splits = ["train", "valid", "test"]
```

### Dismir

```python
from syto.modelling.wrappers import DismirMLflowExperiment

experiment = DismirMLflowExperiment(
    data_path=data_path,
    experiment_name=experiment_name,
    max_sequence_length=max_sequence_length,
    model_flavor="lstm",
    tracking_uri="http://127.0.0.1:5000",
    splits=splits,
)

# Run the complete experiment
results = experiment.run_full_experiment(
    epochs=500,
    batch_size=256,
    patience=12,
    optimizer_type="SGD",
    lr=0.0025,
    momentum=0.9,
)
print("Experiment completed!")
```

### DNABERT2

```python
# data_path = "../Data/Curated/rrms_dmrs_only_mincov40_corrected_top3000_dmr_cuts/"
from syto.modelling.wrappers import EpigenBERT2MLflowExperiment
from syto.modelling.dnabert2 import TrainingArguments
```

```python
training_args = TrainingArguments(
    run_name="dnabert2_" + data_path.split("/")[-1],
    per_device_train_batch_size=100,
    per_device_eval_batch_size=200,
    gradient_accumulation_steps=1,
    learning_rate=3e-5,
    fp16=True,
    save_steps=10,
    eval_strategy="steps",
    eval_steps=10,
    warmup_steps=100,
    logging_steps=100,
    num_train_epochs=100,
    overwrite_output_dir=True,
    log_level="info",
    find_unused_parameters=False,
    batch_eval_metrics=False,
    eval_and_save_results=True,
    remove_unused_columns=False,
    eval_accumulation_steps=8,
    torch_empty_cache_steps=10,
    prediction_loss_only=False,
    gradient_checkpointing=False,
    skip_memory_metrics=True,
    auto_find_batch_size=False,
)
```

```python
experiment = EpigenBERT2MLflowExperiment(
    data_path=data_path,
    experiment_name=experiment_name,
    max_sequence_length=max_sequence_length,
    tracking_uri="http://127.0.0.1:5000",
    splits=splits,
)

# Run the complete experiment
results = experiment.run_full_experiment(
    training_args,
    **{
        "early_stopping_patience": 12,
        "eval_loss_threshold": 0.55,
        "check_at_step": 400,
        "max_retries": 10,
    }
)
print("Experiment completed!")
```
