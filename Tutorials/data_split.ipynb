```python
from sklearn.model_selection import StratifiedShuffleSplit
import pandas as pd
```

```python
data_path = "../Data/Curated/MethylBERT_Figure3D"
save_path = "../Data/Curated/MethylBERT_Figure3D_customsplit"
data = pd.read_csv(data_path+"/train.csv")
```

```python
sss = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=1238594)
sss.split(data[["input_ids", "methylation_ids"]], data[["label"]])

for i, (train_index, test_index) in enumerate(sss.split(data[["input_ids", "methylation_ids"]], data[["label"]])):
        print(f"Fold {i}:")
        print(f"  Train: index={train_index}")
        print(f"  Test:  index={test_index}")
```

```python
data.iloc[train_index].to_csv(save_path+"/train.csv", index = False)
data.iloc[test_index].to_csv(save_path+"/valid.csv", index = False)
```
