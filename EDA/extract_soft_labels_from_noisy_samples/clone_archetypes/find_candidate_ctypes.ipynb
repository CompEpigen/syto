```python
from pathlib import Path
import pandas as pd
```

```python
PATH_TO_DATA = Path(
    "/staging/leuven/stg_00118/methylDL/data/loyfer2023/hg38/data/GSE186458"
)
unique_ctypes = set()
for file in PATH_TO_DATA.glob("*.pat.gz"):
    filename = file.stem
    ctype = filename.split("_")[1].split(".")[0]
    ctype = ctype[::-1].split("-", 1)[-1][::-1]
    if "CNVS" in ctype:
        continue
    unique_ctypes.add(ctype)
```
