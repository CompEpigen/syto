```python
metrics = {
    "dismir": {
        "hard_labels": {
            "diagrej": metrics_dismir_hardlabels_diagrej,
            "top156": metrics_dismir_hardlabels_top156,
        },
        "canonical_soft_labels": {
            "diagrej": metrics_dismir_canonicalsoft_diagrej,
            "top156": metrics_dismir_canonicalsoft_top156,
        },
        "soft_labels_without_pooling": metrics_dismir_softlabels_without_pooling,
        "soft_labels_with_pooling": metrics_dismir_softlabels_with_pooling,
    },
    "methylbert": {
        "hard_labels": {
            "diagrej": metrics_methylbert_hardlabels_diagrej,
            "top156": metrics_methylbert_hardlabels_top156,
        },
        "canonical_soft_labels": {
            "diagrej": metrics_methylbert_canonicalsoft_diagrej,
            "top156": metrics_methylbert_canonicalsoft_top156,
        },
        "soft_labels_without_pooling": metrics_methylbert_softlabels_without_pooling,
        "soft_labels_with_pooling": metrics_methylbert_softlabels_with_pooling,
    },
    "lookup": {
        "hard_labels": {
            "diagrej": metrics_lookup_hardlabels_diagrej,
            "top156": metrics_lookup_hardlabels_top156,
        },
        "soft_labels_without_pooling": metrics_lookup_softlabels_without_pooling,
        "soft_labels_with_pooling": metrics_lookup_softlabels_with_pooling,
    },
    "cancerdetector": {
        "train_freq_prior": metrics_cancerdetector_trainfreq,
        "uniform_prior": metrics_cancerdetector_uniform,
    },
    "uxm": metrics_uxm,
}
```

```python
COL_CRITERION = "kl"
```

```python
# first we select the best pair of (deconvolver, calibration method) for each (classifier, labeling)
best_metrics = {}
for clf in ["dismir", "methylbert", "lookup"]:
    best_metrics[clf] = {"cal": {}, "uncal": {}}
    for labeling_type in metrics[clf].keys():
        if labeling_type in ["hard_labels", "canonical_soft_labels"]:
            best_metrics[clf]["cal"][labeling_type] = {}
            best_metrics[clf]["uncal"][labeling_type] = {}
            for feature_set in ["diagrej", "top156"]:
                df = metrics[clf][labeling_type][feature_set]
                best_row = df.loc[df[COL_CRITERION].idxmin()]
                best_metrics[clf]["cal"][labeling_type][feature_set] = best_row.to_dict()

                df_uncal = df[df["calibration_method"] == "uncalibrated"]
                best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
                best_metrics[clf]["uncal"][labeling_type][feature_set] = best_row_uncal.to_dict()
        else:
            df = metrics[clf][labeling_type]
            best_row = df.loc[df[COL_CRITERION].idxmin()]
            best_metrics[clf]["cal"][labeling_type] = best_row.to_dict()
            
            df_uncal = df[df["calibration_method"] == "uncalibrated"]
            best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
            best_metrics[clf]["uncal"][labeling_type] = best_row_uncal.to_dict()
```

```python
best_metrics["cancerdetector"] = {"cal": {}, "uncal": {}}
for prior in ["train_freq_prior", "uniform_prior"]:
    df = metrics["cancerdetector"][prior]
    best_row = df.loc[df[COL_CRITERION].idxmin()]
    best_metrics["cancerdetector"]["cal"][prior] = best_row.to_dict()

    df_uncal = df[df["calibration_method"] == "uncalibrated"]
    best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
    best_metrics["cancerdetector"]["uncal"][prior] = best_row_uncal.to_dict()
```

```python
## print the table
classifiers_print = {
    "dismir": "{\\shortstack[l]{Dismir\\cite{li2021dismir}\\\\(CNN+LSTM)}}",
    "methylbert": "MethylBERT \\cite{jeong2025methylbert}",
    "lookup": "\\shortstack[l]{Lookup Classifier\\\\(1-NN)}"
}
labeling_print = {
    "hard_labels": "Hard labels w/ bckg.",
    "canonical_soft_labels": "Canonical Soft Labels",
    "soft_labels_without_pooling": "DD Soft Labels w/o pool.",
    "soft_labels_with_pooling": "DD Soft Labels w/ pool."
}
deconvolvers_print = {
    "nnls": "NNLS",
    "psls": "PSLS",
    "swn": "SWN",
    "xgb": "XGB",
    "mlp": "MLP",
}
prior_print = {
    "train_freq_prior": "Prior: Training counts",
    "uniform_prior": "Prior: Uniform",
}
calibrator_print = {
    "uncalibrated": "Uncalibrated",
    "linear_clip0_normalize": r"Lin.\ clip+norm",
    "linear_simplex_projection": r"Lin.\ simplex",
    "vector_scaling": r"Vec.\ scaling",
}
```

```python
# first we print the uncalibrated section
for i_clf, clf in enumerate(["dismir", "methylbert", "lookup"]):
    print("\\multirow{" + str(len(metrics[clf])) + "}{*}{" + classifiers_print[clf] + "}", end="\n")
    for i_labeling, labeling in enumerate(metrics[clf].keys()):
        # extract the metrics to print on the line
        if labeling in ["hard_labels", "canonical_soft_labels"]:
            metric_diagrej = best_metrics[clf]["uncal"][labeling]["diagrej"][COL_CRITERION]
            metric_top156 = best_metrics[clf]["uncal"][labeling]["top156"][COL_CRITERION]
            is_diagrej_better = metric_diagrej > metric_top156
            if is_diagrej_better:
                line_metrics = best_metrics[clf]["uncal"][labeling]["diagrej"]
            else:
                line_metrics = best_metrics[clf]["uncal"][labeling]["top156"]   
        else:
            line_metrics = best_metrics[clf]["uncal"][labeling]
        line = [" & " + labeling_print[labeling]]
        line.append(
            deconvolvers_print[line_metrics["deconvolver"]]
            + (r"\textsuperscript{\textdagger}" if is_diagrej_better else "")
        )
        line.append("None")
        # r2
        line.append(f"{line_metrics['r2']*1e2:.2f} {{\\scriptsize[{line_metrics['r2_ci_lower']*1e2:.2f}, {line_metrics['r2_ci_upper']*1e2:.2f}]}}")
        # mae
        line.append(f"{line_metrics['mae']*1e3:.2f} {{\\scriptsize[{line_metrics['mae_ci_lower']*1e3:.2f}, {line_metrics['mae_ci_upper']*1e3:.2f}]}}")
        # mse
        line.append(f"{line_metrics['mse']*1e4:.2f} {{\\scriptsize[{line_metrics['mse_ci_lower']*1e4:.2f}, {line_metrics['mse_ci_upper']*1e4:.2f}]}}")
        # kl
        line.append(f"{line_metrics['kl']*1e2:.2f} {{\\scriptsize[{line_metrics['kl_ci_lower']*1e2:.2f}, {line_metrics['kl_ci_upper']*1e2:.2f}]}}")
        line_str =" & ".join(line) + "\\\\"
        print(line_str)
    if i_clf < 2:
        print("\\cmidrule(l){2-8}")
```
