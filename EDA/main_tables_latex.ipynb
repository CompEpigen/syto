## Set print settings

```python
COL_CRITERION = "mse"
DO_PRINT_METRIC = {
    "r2": True,
    "mae": True,
    "mse": True,
    "kl": False,
}
NUM_COLUMNS = 4 + sum(int(v) for v in DO_PRINT_METRIC.values())
```

## Selection of best deconvolver and calibrator for each pair (classifier + labeling/prior)

```python
# first we select the best pair of (deconvolver, calibration method) for each (classifier, labeling)
best_metrics = {}
for clf in ["dismir", "methylbert", "lookup"]:
    best_metrics[clf] = {"cal": {}, "uncal": {}}
    for labeling_type in metrics[clf].keys():
        if labeling_type in ["hard_labels", "canonical_soft_labels"]:
            best_metrics[clf]["cal"][labeling_type] = {}
            best_metrics[clf]["uncal"][labeling_type] = {}
            for feature_set in ["diagbckg", "top156"]:
                df = metrics[clf][labeling_type][feature_set]
                best_row = df.loc[df[COL_CRITERION].idxmin()]
                best_metrics[clf]["cal"][labeling_type][
                    feature_set
                ] = best_row.to_dict()

                df_uncal = df[df["calibration_method"] == "uncalibrated"]
                best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
                best_metrics[clf]["uncal"][labeling_type][
                    feature_set
                ] = best_row_uncal.to_dict()
        else:
            df = metrics[clf][labeling_type]
            best_row = df.loc[df[COL_CRITERION].idxmin()]
            best_metrics[clf]["cal"][labeling_type] = best_row.to_dict()

            df_uncal = df[df["calibration_method"] == "uncalibrated"]
            best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
            best_metrics[clf]["uncal"][labeling_type] = best_row_uncal.to_dict()
```

```python
best_metrics["cancerdetector"] = {"cal": {}, "uncal": {}}
for prior in ["train_freq_prior", "uniform_prior"]:
    df = metrics["cancerdetector"][prior]
    best_row = df.loc[df[COL_CRITERION].idxmin()]
    best_metrics["cancerdetector"]["cal"][prior] = best_row.to_dict()

    df_uncal = df[df["calibration_method"] == "uncalibrated"]
    best_row_uncal = df_uncal.loc[df_uncal[COL_CRITERION].idxmin()]
    best_metrics["cancerdetector"]["uncal"][prior] = best_row_uncal.to_dict()
```

## Find lines for bold, italic, underline, highlighted

Independently for each metric:
- results of Best labeling/prior for each classifier is italicized
- results of best per section is bolded
- results of second best per section is underlined
- results of best across all sections is highlighted 

```python
def dict_arg_min_max(d, maximize=True):
    best_key = None
    best_value = -np.inf if maximize else np.inf
    for k, v in d.items():
        if maximize and v >= best_value:
            best_key = k
            best_value = v
        if (not maximize) and v <= best_value:
            best_key = k
            best_value = v
    return best_key


def dict_snd_arg_min_max(d, maximize=True):
    best_key = dict_arg_min_max(d, maximize=maximize)
    copy_d = d.copy()
    del copy_d[best_key]
    return dict_arg_min_max(copy_d, maximize=maximize)
```

```python
best_feature_set = {}
for clf in ["dismir", "methylbert", "lookup"]:
    best_feature_set[clf] = {}
    for cal_type in ["cal", "uncal"]:
        best_feature_set[clf][cal_type] = {}
        for labeling_type in best_metrics[clf][cal_type].keys():
            if labeling_type in ["hard_labels", "canonical_soft_labels"]:
                best_feature_set[clf][cal_type][labeling_type] = {}
                diagbckg_mse = best_metrics[clf][cal_type][labeling_type]["diagbckg"][
                    COL_CRITERION
                ]
                top156_mse = best_metrics[clf][cal_type][labeling_type]["top156"][
                    COL_CRITERION
                ]
                if diagbckg_mse < top156_mse:
                    best_feature_set[clf][cal_type][labeling_type] = "diagbckg"
                else:
                    best_feature_set[clf][cal_type][labeling_type] = "top156"
            else:
                best_feature_set[clf][cal_type][labeling_type] = None
```

```python
should_italicize = {}
for clf in ["dismir", "methylbert", "lookup"]:
    should_italicize[clf] = {}
    for cal_type in ["uncal", "cal"]:
        should_italicize[clf][cal_type] = {}
        for metric_name in METRICS:
            should_italicize[clf][cal_type][metric_name] = dict_arg_min_max(
                {
                    labeling_type: (
                        best_metrics[clf][cal_type][labeling_type][metric_name]
                        if best_feature_set[clf][cal_type][labeling_type] is None
                        else best_metrics[clf][cal_type][labeling_type][
                            best_feature_set[clf][cal_type][labeling_type]
                        ][metric_name]
                    )
                    for labeling_type in best_metrics[clf][cal_type].keys()
                },
                maximize=metric_name in ["r2"],
            )
```

```python
should_bold = {}
should_underline = {}
for cal_type in ["uncal", "cal"]:
    should_bold[cal_type] = {}
    should_underline[cal_type] = {}
    for metric_name in METRICS:
        maximize = metric_name in ["r2"]
        tmp_dict = {
            (clf, labeling_prior): (
                best_metrics[clf][cal_type][labeling_prior][metric_name]
                if clf == "cancerdetector"
                or best_feature_set[clf][cal_type][labeling_prior] is None
                else best_metrics[clf][cal_type][labeling_prior][
                    best_feature_set[clf][cal_type][labeling_prior]
                ][metric_name]
            )
            for clf in ["dismir", "methylbert", "lookup", "cancerdetector"]
            for labeling_prior in best_metrics[clf][cal_type].keys()
        }
        should_bold[cal_type][metric_name] = dict_arg_min_max(
            tmp_dict,
            maximize=maximize,
        )
        should_underline[cal_type][metric_name] = dict_snd_arg_min_max(
            tmp_dict,
            maximize=maximize,
        )
# now for the baselines
should_bold["baselines"] = {}
should_underline["baselines"] = {}
for metric_name in METRICS:
    maximize = metric_name in ["r2"]
    tmp_dict = {
        (baseline, cal_type): best_metrics[baseline][cal_type][metric_name]
        for baseline in best_metrics.keys()
        if baseline not in ["dismir", "methylbert", "lookup", "cancerdetector"]
        for cal_type in best_metrics[baseline].keys()
    }
    should_bold["baselines"][metric_name] = dict_arg_min_max(
        tmp_dict,
        maximize=maximize,
    )
    should_underline["baselines"][metric_name] = dict_snd_arg_min_max(
        tmp_dict,
        maximize=maximize,
    )
```

```python
should_highlight = {}
for metric in METRICS:
    maximize = metric in ["r2"]
    tmp_dict = {
        (clf, labeling_prior, cal_type): (
            best_metrics[clf][cal_type][labeling_prior][metric]
            if clf == "cancerdetector"
            or best_feature_set[clf][cal_type][labeling_prior] is None
            else best_metrics[clf][cal_type][labeling_prior][
                best_feature_set[clf][cal_type][labeling_prior]
            ][metric]
        )
        for clf in ["dismir", "methylbert", "lookup", "cancerdetector"]
        for cal_type in ["uncal", "cal"]
        for labeling_prior in best_metrics[clf][cal_type].keys()
    }
    should_highlight[metric] = dict_arg_min_max(
        tmp_dict,
        maximize=maximize,
    )
```

```python
# first we print the uncalibrated section
for i_clf, clf in enumerate(["dismir", "methylbert", "lookup"]):
    print(
        r"\multirow{"
        + str(len(metrics[clf]))
        + "}{*}{"
        + classifiers_print[clf]
        + r"}",
    )
    for i_labeling, labeling in enumerate(metrics[clf].keys()):
        # extract the metrics to print on the line
        current_best_feature_set = best_feature_set[clf]["uncal"][labeling]
        if labeling in ["hard_labels", "canonical_soft_labels"]:
            line_metrics = best_metrics[clf]["uncal"][labeling][
                current_best_feature_set
            ]
        else:
            line_metrics = best_metrics[clf]["uncal"][labeling]
        line = [" & " + labeling_print[labeling]]
        line.append(
            deconvolvers_print[line_metrics["deconvolver"]]
            + (
                r"\textsuperscript{\textdagger}"
                if current_best_feature_set == "diagbckg"
                else ""
            )
        )
        line.append("None")
        for metric in METRICS:
            if DO_PRINT_METRIC[metric]:
                metric_value = line_metrics[metric]
                metric_ci_lower = line_metrics[f"{metric}_ci_lower"]
                metric_ci_upper = line_metrics[f"{metric}_ci_upper"]
                multiplier = METRIC_MULTIPLIER[metric]
                core_string = f"{metric_value*multiplier:.2f} {{\\scriptsize[{metric_ci_lower*multiplier:.2f}, {metric_ci_upper*multiplier:.2f}]}}"
                if should_bold["uncal"][metric] == (clf, labeling):
                    core_string = r"\textbf{" + core_string + "}"
                if should_italicize[clf]["uncal"][metric] == labeling:
                    core_string = r"\textit{" + core_string + "}"
                if should_underline["uncal"][metric] == (clf, labeling):
                    core_string = r"\underline{" + core_string + "}"
                if should_highlight[metric] == (clf, labeling, "uncal"):
                    core_string = r"\cellcolor{teal!27}" + core_string

                line.append(core_string)
        line_str = " & ".join(line) + "\\\\"
        print(line_str)
    if i_clf < 2:
        print(rf"\cmidrule(l){{2-{NUM_COLUMNS}}}")
# cancer detector uncalibrated
print(rf"\cmidrule(l){{2-{NUM_COLUMNS}}}")
print(r"\multirow{2}{*}{CancerDetector\cite{li2018cancerdetector}}")
clf = "cancerdetector"
for prior in metrics["cancerdetector"].keys():
    line_metrics = best_metrics["cancerdetector"]["uncal"][prior]
    line = [" & " + prior_print[prior]]
    line.append(deconvolvers_print[line_metrics["deconvolver"]])
    line.append("None")
    for metric in METRICS:
        if DO_PRINT_METRIC[metric]:
            metric_value = line_metrics[metric]
            metric_ci_lower = line_metrics[f"{metric}_ci_lower"]
            metric_ci_upper = line_metrics[f"{metric}_ci_upper"]
            multiplier = METRIC_MULTIPLIER[metric]
            core_string = f"{metric_value*multiplier:.2f} {{\\scriptsize[{metric_ci_lower*multiplier:.2f}, {metric_ci_upper*multiplier:.2f}]}}"
            if should_bold["uncal"][metric] == (clf, prior):
                core_string = r"\textbf{" + core_string + "}"
            if should_underline["uncal"][metric] == (clf, prior):
                core_string = r"\underline{" + core_string + "}"
            if should_highlight[metric] == (clf, prior):
                core_string = r"\cellcolor{teal!27}" + core_string

            line.append(core_string)
    line_str = " & ".join(line) + r"\\"
    print(line_str)
```
